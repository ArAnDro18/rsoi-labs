1. Проверяем базовые функции сервера

```python
# xmlrpc_client.ipynb

import xmlrpc.client as xmlrpclib
import pickle
import pandas as pd

server = xmlrpclib.ServerProxy("http://127.0.0.1:8009", allow_none=True)

```

```python
print ('Ping:', server.ping())

print ('Server datetime:', server.now())

print ('View, type, value:', server.type(2))
print ('View, type, value:', server.type(2.))
print ('View, type, value:', server.type('My string'))
print ('View, type, value:', server.type("My string"))
print ('View, type, value:', server.type([1,2,3]))
print ('View, type, value:', server.type(["one", "two", "three"]))
print ('View, type, value:', server.type((1,2,"3")))

print ('Sum 2 + 3 :', server.sum(2, 3))
print ('Pow 2^3: ', server.pow(2, 3))
```

```python
print(server.get_log())
```

```python
print(server.get_log("ping"))
```

```python
# Все вызовы за указанный период.
print(server.get_log(None,
    "2026-10-19 16:17:40",
    "2026-10-19 16:20:38"
))
```

```python
# Все вызовы длительностью от 0.1 секунды.
print(server.get_log(
    None, None, None,
    0.016, None
))
```

2. Работаем с первым изображением

```python
from PIL import Image # Работа с изображением
from pylab import * # Отображение

import xmlrpc.client as xmlrpclib

# Режим отображения внутри ноутбука
%matplotlib inline

img = Image.open('Jellyfish.jpg')

img_arr = array(img)

# Исходное изображение
imshow(img_arr)
```

```python
pimg = pickle.dumps(img_arr) # универсально

img_bin = xmlrpclib.Binary(pimg)

img_bin2 = server.send_back_binary(img_bin)

#img_arr2 = np.loads(img_bin2.data)
img_arr2 = pickle.loads(img_bin2.data) # универсально

# Изображение после возрата с сервера
imshow(img_arr2)
```

!!! Использование базовой функции инверсии !!!

```python
def inv_color(img_arr_in):
    pimg = pickle.dumps(img_arr_in)
    img_bin = xmlrpclib.Binary(pimg)

    img_bin2 = server.send_back_inversion(img_bin)

    img_arr_out = pickle.loads(img_bin2.data)
    return img_arr_out

img_arr_inv = inv_color(img_arr)

# Изображение с инверсией цвета
imshow(img_arr_inv)
```

Бинаризация

```python
def bin_color(img_arr_in, threshold):
    pimg = pickle.dumps(img_arr_in)
    img_bin = xmlrpclib.Binary(pimg)  

    # Вызов функции binarize на сервере с передачей порога 
    img_bin2 = server.binarize(img_bin, threshold)

    img_arr_out = pickle.loads(img_bin2.data)
    return img_arr_out

# Вызов пороговой бинаризации
img_arr_bin = bin_color(img_arr, 80)

# Изображение после бинаризации
imshow(img_arr_bin)
```

```python
def flip_vert(img_arr_in):
    pimg = pickle.dumps(img_arr_in)
    img_bin = xmlrpclib.Binary(pimg)  # Если Python 2, измените на xmlrpclib.Binary(pimg)

    # Вызов функции flip_vertical на сервере
    img_bin2 = server.flip_vertical(img_bin)

    img_arr_out = pickle.loads(img_bin2.data)
    return img_arr_out

# Вызов разворота изображения
img_arr_flip = flip_vert(img_arr)

# Изображение после разворота по вертикали
imshow(img_arr_flip)
```

Работаем с ЧБ изображением

```python
from PIL import Image # Работа с изображением
from pylab import * # Отображение

import xmlrpc.client as xmlrpclib

# Режим отображения внутри ноутбука
%matplotlib inline

img = Image.open('11.bmp')

img_arr = array(img)

# Исходное изображение
imshow(img_arr, cmap='gray', vmin=0, vmax=255)
```

```python
pimg = pickle.dumps(img_arr) # универсально

img_bin = xmlrpclib.Binary(pimg)

img_bin2 = server.send_back_binary(img_bin)

#img_arr2 = np.loads(img_bin2.data)
img_arr2 = pickle.loads(img_bin2.data) # универсально

# Изображение после возрата с сервера
imshow(img_arr2, cmap='gray', vmin=0, vmax=255)
```

```python
def inv_color(img_arr_in):
    pimg = pickle.dumps(img_arr_in)
    img_bin = xmlrpclib.Binary(pimg)

    img_bin2 = server.back_inversion(img_bin)

    img_arr_out = pickle.loads(img_bin2.data)
    return img_arr_out

img_arr_inv = inv_color(img_arr)

# Изображение с инверсией цвета
imshow(img_arr_inv, cmap='gray', vmin=0, vmax=255)
```
